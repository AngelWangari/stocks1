```python
import yfinance as yf
import pandas as pd

# 1. Create Amazon ticker object
ticker_obj = yf.Ticker('AMZN')

# 2. Get Amazon earnings dates
earnings = ticker_obj.get_earnings_dates()

# 3. Remove rows where earnings surprise is missing
earnings = earnings.dropna(
    subset=['Surprise(%)']
)

# 4. Make earnings dates consistent
earnings.index = pd.to_datetime(
    earnings.index,
    utc=True
).tz_localize(None).normalize()

# 5. Download Amazon historical prices
amzn = yf.download(
    'AMZN',
    start='2020-01-01',
    end='2026-09-09',
    auto_adjust=False
)

# 6. Get closing prices
close = amzn['Close']['AMZN']

# 7. Calculate 2-day return
two_day_return = close.shift(-2) / close - 1

# 8. Create DataFrame containing returns
returns_df = pd.DataFrame({
    '2-Day Return': two_day_return
})

# 9. Make stock-price dates consistent
returns_df.index = pd.to_datetime(
    returns_df.index
).tz_localize(None).normalize()

# 10. Join earnings surprise with stock returns
combined = earnings[['Surprise(%)']].join(
    returns_df[['2-Day Return']],
    how='inner'
)

# 11. Keep only positive earnings surprises
positive = combined[
    combined['Surprise(%)'] > 0
]

# 12. Calculate median 2-day return
median_return = positive['2-Day Return'].median()

# 13. Calculate correlation
correlation = positive['Surprise(%)'].corr(
    positive['2-Day Return']
)

# 14. Print results
print("Median 2-day return:",
      median_return * 100, "%")

print("Correlation:",
      correlation)
```

```python
tickers = {
    'USA': '^GSPC',
    'China': '000001.SS',
    'Hong Kong': '^HSI',
    'Australia': '^AXJO',
    'India': '^NSEI',
    'Canada': '^GSPTSE',
    'Germany': '^GDAXI',
    'UK': '^FTSE',
    'Japan': '^N225',
    'Mexico': '^MXX',
    'Brazil': '^BVSP'
}
```

```python
results = []

for country, ticker in tickers.items():

    data = yf.download(
        ticker,
        start='2026-01-01',
        end='2026-08-22',
        auto_adjust=False
    )

    close = data['Close'].squeeze()

    start_price = close.iloc[0]
    end_price = close.iloc[-1]

    ytd_return = (
        (end_price - start_price)
        / start_price
        * 100
    )

    results.append({
        'Country': country,
        'Ticker': ticker,
        'Start Price': start_price,
        'End Price': end_price,
        'YTD Return %': ytd_return
    })
```

```python
results_df = pd.DataFrame(results)
```

```python
us_return = results_df.loc[
    results_df['Country'] == 'USA',
    'YTD Return %'
].iloc[0]
```

```python
foreign_indexes = results_df[
    results_df['Country'] != 'USA'
]
```

```python
better_than_us = foreign_indexes[
    foreign_indexes['YTD Return %'] > us_return
]
```
